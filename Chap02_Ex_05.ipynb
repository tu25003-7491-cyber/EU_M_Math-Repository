```python
import numpy as np
import numpy.linalg as LA

A = np.array([[1, 2, 3],[1, 3, 2], [3, 1, 2]])
A_inv = LA.inv(A)
print("Aの逆行列:")
print(A_inv)

eigenvalues, eigenvectors = LA.eig(A)

print("固有値:")
print(eigenvalues)

print("固有ベクトル")
print(eigenvectors)
```

```python
eigenvalues, eigenvectors = LA.eig(A)

print("\nAの固有値:")
print(eigenvalues)

print("\nAの固有ベクトル (各列が固有ベクトル):")
print(eigenvectors)
```
