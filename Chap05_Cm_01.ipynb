```python
import numpy as np
from scipy import linalg

A = np.array([
    [5, 1, 0, 1],
    [1, 9, -5, 7],
    [0, -5, 8, -3],
    [1, 7, -3, 10]
], dtype=float)

b = np.array([2, 10, 5, 10], dtype=float)

# コレスキー分解 A = L L^T
L = linalg.cholesky(A, lower=True)

# L y = b を解く
y = linalg.solve_triangular(L, b, lower=True)

# L^T x = y を解く
x = linalg.solve_triangular(L.T, y, lower=False)

print("L =")
print(L)

print("x =")
print(x)

print("確認 A x =")
print(A @ x)
```
