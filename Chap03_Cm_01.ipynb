```python
import pandas as pd
import matplotlib.pyplot as plt

# CSV読み込み
df = pd.read_csv('/content/sample_data/student-mat.csv', sep=';')

# 要約統計量
summary = df.describe()
print(summary)

# CSV保存
summary.to_csv('/content/sample_data/summary_statistics.csv')

# 数値データだけ取り出す
num_df = df.select_dtypes(include='number')

# 分散
variance = num_df.var()
print(variance)

# 分散をCSV保存
variance.to_csv('/content/sample_data/variance.csv', header=['variance'])

# グラフに使う変数を絞る
cols = ['age', 'studytime', 'failures', 'absences', 'G1', 'G2', 'G3']
plot_df = num_df[cols]

# 分散の棒グラフ
plot_df.var().plot(kind='bar')
plt.title('Variance of Variables')
plt.ylabel('Variance')
plt.tight_layout()
plt.show()

# 2変数間の関係を見る散布図行列
pd.plotting.scatter_matrix(plot_df, figsize=(10, 10), diagonal='hist')
plt.suptitle('Scatter Matrix', y=1.02)
plt.tight_layout()
plt.show()
```
