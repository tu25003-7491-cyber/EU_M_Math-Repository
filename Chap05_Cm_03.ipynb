```python
import numpy as np
from scipy import optimize

# 目的関数
def f(x):
    return x[0]**3 + 1

# 制約条件 x >= -1
constraints = [
    {
        "type": "ineq",
        "fun": lambda x: x[0] + 1
    }
]

# 初期値
x0 = np.array([-0.5])

result = optimize.minimize(
    f,
    x0,
    constraints=constraints,
    method="SLSQP"
)

print("最適解 x =", result.x)
print("最小値 f(x) =", result.fun)
print("成功したか =", result.success)
print("メッセージ =", result.message)
```
