```python
import numpy as np
import scipy as sp
import pandas as pd
from pandas import Series, DataFrame
from scipy import stats

# CSVの読み込み
df_mat = pd.read_csv('/content/sample_data/student-mat.csv', sep=';')
df_por = pd.read_csv('/content/sample_data/student-por.csv', sep=';')

# マージ処理（変数名と引数を修正）
student_data_merge = pd.merge(
    df_mat,
    df_por,
    on=['school', 'sex', 'age', 'address', 'famsize', 'Pstatus', 'Medu', 'Fedu', 'Mjob', 'Fjob', 'reason', 'nursery', 'internet'],
    suffixes=('_math', '_por')
)

# 対応ありt検定を行う関数
def paired_test(col):
    math_col = f'{col}_math'
    por_col = f'{col}_por'

    math_data = student_data_merge[math_col]
    por_data = student_data_merge[por_col]

    t_stat, p_value = stats.ttest_rel(math_data, por_data)

    print(f'--- {col} ---')
    print(f'数学の平均: {math_data.mean():.3f}')
    print(f'ポルトガル語の平均: {por_data.mean():.3f}')
    print(f'平均差（数学 - ポルトガル語）: {(math_data - por_data).mean():.3f}')
    print(f't値: {t_stat:.3f}')
    print(f'p値: {p_value:.5f}')

    print()


# 1. 欠席数の差
paired_test('absences')

# 2. 勉強時間の差
paired_test('studytime')

```
