```python
import numpy as np
import scipy as sp
import pandas as pd
from pandas import Series, DataFrame
from scipy import stats



df_mat = pd.read_csv('/content/sample_data/student-mat.csv', sep=';')
#print(df_mat)
df_por = pd.read_csv('/content/sample_data/student-por.csv', sep=';')
#print(df_por)

from scipy import stats

def compare_subject_mean(col):
    mat = df_mat[col].dropna()
    por = df_por[col].dropna()

    t_stat, p_value = stats.ttest_ind(mat, por, equal_var=False)

    return {
        "成績": col,
        "数学の平均": mat.mean(),
        "ポルトガル語の平均": por.mean(),
        "平均差(数学-ポルトガル語)": mat.mean() - por.mean(),
        "t値": t_stat,
        "p値": p_value,
        "5%水準で差あり": p_value < 0.05
    }

result = pd.DataFrame([
    compare_subject_mean("G2"),
    compare_subject_mean("G3")
])

print(result)
```
