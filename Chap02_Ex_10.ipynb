```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-10, 10, 400) # Generate 400 points between -10 and 10
y = 5 * x + 3

plt.figure(figsize=(8, 6))
plt.plot(x, y, label='y = 5x + 3')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axhline(0, color='black', linewidth=0.5) # Add x-axis line
plt.axvline(0, color='black', linewidth=0.5) # Add y-axis line
plt.legend()
plt.show()
```
