```python
import numpy as np
import matplotlib.pyplot as plt
import math

# 1. 一様乱数を2組作る

n = 10000

x = np.random.uniform(0.0, 1.0, n)
y = np.random.uniform(0.0, 1.0, n)
```

```python
# 2　円の中に入る点を調べる
distance = np.hypot(x, y)

inside = distance <= 1.0
outside = distance > 1.0

inside_count = np.sum(inside)

print("円の中に入った点の数:", inside_count)
print("全体の点の数:", n)

plt.figure(figsize=(6, 6))

plt.scatter(x[inside], y[inside], s=5, label="inside")
plt.scatter(x[outside], y[outside], s=5, label="outside")

plt.xlim(0, 1)
plt.ylim(0, 1)
plt.gca().set_aspect("equal")

plt.title("Monte Carlo Simulation")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# 3　円周率を求める

pi = 4 * inside_count / n

print("推定した円周率:", pi)
print("実際の円周率:", math.pi)
```
