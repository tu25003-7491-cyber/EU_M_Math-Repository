```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-10, 10, 400) # Generate 400 points between -10 and 10
y_sin = np.sin(x)
y_cos = np.cos(x)

plt.figure(figsize=(10, 6))
plt.plot(x, y_sin, label='y = sin(x)', color='blue')
plt.plot(x, y_cos, label='y = cos(x)', color='red')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axhline(0, color='black', linewidth=0.5) # Add x-axis line
plt.axvline(0, color='black', linewidth=0.5) # Add y-axis line
plt.legend()
plt.show()
```
