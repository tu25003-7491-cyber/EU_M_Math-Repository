```python
import numpy as np
import matplotlib.pyplot as plt

# 乱数の個数を変えて試す
sample_sizes = [100, 1000, 10000]

for n in sample_sizes:
    # 0〜1の一様乱数を2組発生
    data1 = np.random.uniform(0.0, 1.0, n)
    data2 = np.random.uniform(0.0, 1.0, n)

    # グラフ作成
    plt.figure(figsize=(10, 4))

    plt.subplot(1, 2, 1)
    plt.hist(data1, bins=20, edgecolor="black")
    plt.title(f"Uniform Random Numbers 1\nn = {n}")
    plt.xlabel("Value")
    plt.ylabel("Frequency")

    plt.subplot(1, 2, 2)
    plt.hist(data2, bins=20, edgecolor="black")
    plt.title(f"Uniform Random Numbers 2\nn = {n}")
    plt.xlabel("Value")
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()
```
