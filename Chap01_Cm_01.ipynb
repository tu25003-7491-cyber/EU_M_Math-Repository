```python
def is_prime(n: int) -> bool:
    if n < 2:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False

    i = 3
    while i * i <= n:
        if n % i == 0:
            return False
        i += 2

    return True

N = 10
for N in range(1, N + 1):
  if is_prime(N) == True:
    print(f"{N}は素数")
  else:
    print(f"{N}は素数ではない")
```
