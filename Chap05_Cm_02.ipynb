```python
import numpy as np
from scipy import integrate

# 被積分関数
# dblquadでは f(y, x) の順で定義する
def f(y, x):
    return 1 / (np.sqrt(x + y) * (1 + x + y)**2)

# x: 0〜1
# y: 0〜1-x
result, error = integrate.dblquad(
    f,
    0, 1,
    lambda x: 0,
    lambda x: 1 - x
)

print("積分値 =", result)
print("推定誤差 =", error)
```
